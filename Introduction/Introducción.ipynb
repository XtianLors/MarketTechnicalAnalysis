## Iniciar una Serie Temporal con Frecuencia de 1 hr.


- __Inicio__: 2018-01-01 00:00:00 <br>
- __Periodo__: 1 hr<br>
- __Final__: 2018-01-01 02:00:00<br> 
- __Longitud__: 2 hrs<br>

"Una serie temporal también se puede leer como, inicio, cada cuánto tiempo estaremos contando, el final de la serie y la longitud de tiempo que abarca."

Más técnico "Una ventana temporal regular de tres observaciones con frecuencia horaria." [1]

El tiempo transcurrido desde el inicio es de 2 hrs (0-2 hrs) sin embargo los datos registrados son tres. Es decir, en una serie temporal la únidad de información se puede considerar como mínimo tres datos.

Un solo registro se considera un dato flotante, dos registros es una observación asemántica, debido a que si bien, un método de extrapolación funciona con dos datos, 

```python
dti = pd.date_range("2018-01-01", periods=3, freq="h")
dti
```

```python
print(dti)
print(type(dti))
print(str(dti))
print(len(dti))
print('Metods in dti: ', len(dir(dti)))
print(dir(dti)[0:5])
```

```python
for i in dti:
    print(i)
```

## Iniciar una Serie Temporal con Frecuencia de 1 hr.

En el mundo financiero, es importante manejar datos que permitan identificar el horario y la geolocalización del recurso financiero. Esto con la finalidad de obedecer y cohercionar nuestros modelos con las operaciones de los  participantes del mercado (cadenas de suministro y participantes financieros.)

```python
dti = pd.date_range("2018-01-01", periods=3, freq="h")
dti = dti.tz_localize("UTC")
dti
```

```python
print(dti)
print(type(dti))
print(str(dti))
print(len(dti))
print('Metods in dti: ', len(dir(dti)))
print(dir(dti)[0:5])
print()
```

```python
dti.tz_convert("US/Pacific")
```
